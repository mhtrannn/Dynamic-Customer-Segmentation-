```python
import pandas as pd
# Load the cleaned dataset and convert the date format
df = pd.read_csv("online_retail_cleaned CusID.csv")
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

# Get unique invoices per customer to calculate intervals
Invoices = df[['Customer ID', 'Invoice', 'InvoiceDate']].drop_duplicates().sort_values(['Customer ID', 'InvoiceDate'])

# Calculate days between consecutive purchases for the same customer
invoices['PrevDate'] = invoices.groupby('Customer ID')['InvoiceDate'].shift(1)
invoices['Interval'] = (invoices['InvoiceDate'] - invoices['PrevDate']).dt.days

# Drop the first purchase (NaT) and calculate statistical thresholds
intervals = invoices['Interval'].dropna()

print(f"Median Interval: {intervals.median()} days")
print(f"<= 30 days: {(intervals <= 30).mean() * 100:.2f}%")
print(f"<= 60 days: {(intervals <= 60).mean() * 100:.2f}%")
print(f"<= 90 days: {(intervals <= 90).mean() * 100:.2f}%")
print(f"<= 120 days: {(intervals <= 120).mean() * 100:.2f}%")
```
